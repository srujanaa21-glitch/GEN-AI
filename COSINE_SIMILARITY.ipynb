```python
# ============================================================
# COSINE SIMILARITY - COMPLETE GOOGLE COLAB CODE
# ============================================================

# ------------------------------------------------------------
# 1. Define Two Simple Vectors
# ------------------------------------------------------------

vector_A = [2, 4, 1]
vector_B = [1, 3, 5]

print("=" * 60)
print("COSINE SIMILARITY CALCULATION")
print("=" * 60)

print("\nVector A:", vector_A)
print("Vector B:", vector_B)


# ------------------------------------------------------------
# 2. Dot Product - From Scratch
# ------------------------------------------------------------

def dot_product(A, B):
    total = 0
    for a, b in zip(A, B):
        total += a * b
    return total


dp = dot_product(vector_A, vector_B)

print("\n--- Step 1: Dot Product ---")
print("A . B =", dp)


# ------------------------------------------------------------
# 3. Magnitude - From Scratch
# ------------------------------------------------------------

import math

def magnitude(A):
    total = 0
    for a in A:
        total += a ** 2
    return math.sqrt(total)


mag_A = magnitude(vector_A)
mag_B = magnitude(vector_B)

print("\n--- Step 2: Magnitude ---")
print("Magnitude of A:", round(mag_A, 4))
print("Magnitude of B:", round(mag_B, 4))


# ------------------------------------------------------------
# 4. Cosine Similarity - From Scratch
# ------------------------------------------------------------

def cosine_similarity_manual(A, B):
    dp = dot_product(A, B)
    mag_A = magnitude(A)
    mag_B = magnitude(B)

    # Prevent division by zero
    if mag_A == 0 or mag_B == 0:
        raise ValueError("Cosine similarity is undefined for a zero vector.")

    similarity = dp / (mag_A * mag_B)
    return similarity


result_manual = cosine_similarity_manual(vector_A, vector_B)

print("\n--- Step 3: Cosine Similarity (Manual) ---")
print("Cosine Similarity:", round(result_manual, 6))


# ------------------------------------------------------------
# 5. Verify Using NumPy
# ------------------------------------------------------------

import numpy as np

A = np.array(vector_A)
B = np.array(vector_B)

dot = np.dot(A, B)

norm_A = np.linalg.norm(A)
norm_B = np.linalg.norm(B)

result_numpy = dot / (norm_A * norm_B)

print("\n--- Step 4: Cosine Similarity (NumPy) ---")
print("Cosine Similarity:", round(result_numpy, 6))


# ------------------------------------------------------------
# 6. Verify Using scikit-learn
# ------------------------------------------------------------

# Install scikit-learn if necessary
!pip install -q scikit-learn

from sklearn.metrics.pairwise import cosine_similarity

# scikit-learn expects 2D arrays
A_2d = np.array(vector_A).reshape(1, -1)
B_2d = np.array(vector_B).reshape(1, -1)

result_sklearn = cosine_similarity(A_2d, B_2d)[0][0]

print("\n--- Step 5: Cosine Similarity (scikit-learn) ---")
print("Cosine Similarity:", round(result_sklearn, 6))


# ------------------------------------------------------------
# 7. Compare All Three Results
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("COMPARISON")
print("=" * 60)

print("Manual       :", round(result_manual, 6))
print("NumPy        :", round(result_numpy, 6))
print("scikit-learn :", round(result_sklearn, 6))

# Check that all results are approximately equal
assert np.isclose(result_manual, result_numpy)
assert np.isclose(result_manual, result_sklearn)

print("\n✅ All three methods match!")


# ------------------------------------------------------------
# 8. Test Different Examples
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("TESTING DIFFERENT EXAMPLES")
print("=" * 60)

examples = {
    "Identical vectors": ([1, 2, 3], [1, 2, 3]),
    "Opposite vectors": ([1, 2, 3], [-1, -2, -3]),
    "Perpendicular vectors": ([1, 0], [0, 1]),
    "Scaled vectors": ([1, 2, 3], [2, 4, 6])
}

for name, (a, b) in examples.items():
    similarity = cosine_similarity_manual(a, b)
    print(f"{name:30s} -> {similarity:.4f}")


# ------------------------------------------------------------
# 9. Practice Exercise
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("PRACTICE EXERCISE")
print("=" * 60)

A = [1, 1, 1]
B = [5, 5, 5]

practice_result = cosine_similarity_manual(A, B)

print("Vector A:", A)
print("Vector B:", B)
print("Cosine Similarity:", round(practice_result, 4))


# ------------------------------------------------------------
# 10. Final Explanation
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("KEY TAKEAWAYS")
print("=" * 60)

print("""
Cosine Similarity Formula:

                A . B
Cosine = -------------------
         ||A|| × ||B||

1. Dot Product
   = Sum of element-wise multiplication

2. Magnitude
   = Length of the vector

3. Cosine Similarity
   = Dot Product / (Magnitude A × Magnitude B)

Interpretation:

   1.0  -> Same direction / very similar
   0.0  -> Perpendicular / no directional similarity
  -1.0  -> Opposite direction

Cosine similarity focuses on direction rather than vector length.
This makes it useful for comparing text embeddings.
""")

print("=" * 60)
print("✅ PROGRAM COMPLETED SUCCESSFULLY")
print("=" * 60)
```

```python
# ============================================================
# CHROMADB COMPLETE GOOGLE COLAB DEMO
# From Scratch -> Intermediate
# ============================================================

# ------------------------------------------------------------
# 1. INSTALL REQUIRED PACKAGES
# ------------------------------------------------------------
!pip install -q chromadb sentence-transformers

print("✅ Packages installed successfully!")


# ------------------------------------------------------------
# 2. IMPORT LIBRARIES
# ------------------------------------------------------------
import chromadb
from chromadb.utils import embedding_functions

print("✅ Libraries imported successfully!")


# ------------------------------------------------------------
# 3. CREATE AN IN-MEMORY CHROMADB CLIENT
# ------------------------------------------------------------
# Data stored here will disappear when the Colab runtime restarts.

client = chromadb.Client()

print("✅ ChromaDB client created successfully!")


# ------------------------------------------------------------
# 4. CREATE A COLLECTION
# ------------------------------------------------------------
collection = client.create_collection(
    name="my_first_collection"
)

print("✅ Collection created:", collection.name)


# ------------------------------------------------------------
# 5. ADD DOCUMENTS
# ------------------------------------------------------------
# ChromaDB automatically creates embeddings for these documents.

collection.add(
    documents=[
        "Python is a popular programming language for AI.",
        "The weather today is sunny and warm.",
        "Machine learning models learn patterns from data.",
        "I enjoy playing football on weekends.",
        "Deep learning uses neural networks with many layers.",
        "Paris is the capital city of France."
    ],
    ids=[
        "doc1",
        "doc2",
        "doc3",
        "doc4",
        "doc5",
        "doc6"
    ]
)

print("✅ Documents added!")
print("Total documents:", collection.count())


# ------------------------------------------------------------
# 6. SEMANTIC SEARCH
# ------------------------------------------------------------
print("\n" + "=" * 60)
print("SEMANTIC SEARCH")
print("=" * 60)

query = "Tell me about AI and neural networks"

results = collection.query(
    query_texts=[query],
    n_results=3
)

print("\nQuery:", query)
print("\nTop matching documents:")

for doc, distance in zip(results["documents"][0], results["distances"][0]):
    print(f"Distance: {distance:.4f} | {doc}")


# ------------------------------------------------------------
# 7. ANOTHER SEMANTIC SEARCH
# ------------------------------------------------------------
print("\n" + "=" * 60)
print("SECOND SEARCH")
print("=" * 60)

query = "What is the capital of France?"

results = collection.query(
    query_texts=[query],
    n_results=2
)

print("\nQuery:", query)

for doc, distance in zip(results["documents"][0], results["distances"][0]):
    print(f"Distance: {distance:.4f} | {doc}")


# ------------------------------------------------------------
# 8. ADD DOCUMENTS WITH METADATA
# ------------------------------------------------------------
print("\n" + "=" * 60)
print("METADATA FILTERING")
print("=" * 60)

collection.add(
    documents=[
        "The new smartphone has an amazing camera.",
        "Stock prices rose sharply after the earnings report."
    ],
    metadatas=[
        {"category": "technology"},
        {"category": "finance"}
    ],
    ids=[
        "doc7",
        "doc8"
    ]
)

print("✅ Documents with metadata added!")


# ------------------------------------------------------------
# 9. QUERY USING METADATA FILTER
# ------------------------------------------------------------
query = "Tell me about new gadgets"

results = collection.query(
    query_texts=[query],
    n_results=2,
    where={"category": "technology"}
)

print("\nQuery:", query)
print("Filter: category = technology")

for doc, distance in zip(results["documents"][0], results["distances"][0]):
    print(f"Distance: {distance:.4f} | {doc}")


# ------------------------------------------------------------
# 10. HUGGING FACE EMBEDDINGS
# ------------------------------------------------------------
print("\n" + "=" * 60)
print("HUGGING FACE EMBEDDINGS")
print("=" * 60)

# Use Sentence Transformers through ChromaDB
hf_embedder = embedding_functions.SentenceTransformerEmbeddingFunction(
    model_name="all-MiniLM-L6-v2"
)

print("✅ Hugging Face embedding model loaded!")


# ------------------------------------------------------------
# 11. CREATE COLLECTION USING HUGGING FACE EMBEDDINGS
# ------------------------------------------------------------
custom_collection = client.create_collection(
    name="custom_embedding_collection",
    embedding_function=hf_embedder
)

print("✅ Custom embedding collection created!")


# ------------------------------------------------------------
# 12. ADD DOCUMENTS TO CUSTOM COLLECTION
# ------------------------------------------------------------
custom_collection.add(
    documents=[
        "I love dogs.",
        "I like puppies.",
        "The stock market crashed today.",
        "Dogs are wonderful pets.",
        "I enjoy playing with my puppy."
    ],
    ids=[
        "a1",
        "a2",
        "a3",
        "a4",
        "a5"
    ]
)

print("✅ Documents added using Hugging Face embeddings!")


# ------------------------------------------------------------
# 13. SEARCH USING HUGGING FACE EMBEDDINGS
# ------------------------------------------------------------
query = "puppy lovers"

results = custom_collection.query(
    query_texts=[query],
    n_results=3
)

print("\nQuery:", query)
print("\nTop results:")

for doc, distance in zip(results["documents"][0], results["distances"][0]):
    print(f"Distance: {distance:.4f} | {doc}")


# ------------------------------------------------------------
# 14. UPDATE A DOCUMENT
# ------------------------------------------------------------
print("\n" + "=" * 60)
print("UPDATE DOCUMENT")
print("=" * 60)

collection.update(
    ids=["doc4"],
    documents=[
        "I enjoy playing football and basketball on weekends."
    ]
)

print("✅ doc4 updated successfully!")


# ------------------------------------------------------------
# 15. DELETE A DOCUMENT
# ------------------------------------------------------------
print("\n" + "=" * 60)
print("DELETE DOCUMENT")
print("=" * 60)

collection.delete(
    ids=["doc2"]
)

print("✅ doc2 deleted successfully!")
print("Remaining documents:", collection.count())


# ------------------------------------------------------------
# 16. VERIFY UPDATED DOCUMENT
# ------------------------------------------------------------
print("\n" + "=" * 60)
print("VERIFY UPDATE")
print("=" * 60)

results = collection.get(
    ids=["doc4"]
)

print("Document ID:", results["ids"][0])
print("Document:", results["documents"][0])


# ------------------------------------------------------------
# 17. CREATE PERSISTENT CHROMADB DATABASE
# ------------------------------------------------------------
print("\n" + "=" * 60)
print("PERSISTENT STORAGE")
print("=" * 60)

# This saves ChromaDB data to disk.
persistent_client = chromadb.PersistentClient(
    path="/content/chroma_db"
)

persistent_collection = persistent_client.get_or_create_collection(
    name="persistent_demo"
)


# ------------------------------------------------------------
# 18. ADD DATA TO PERSISTENT DATABASE
# ------------------------------------------------------------
# Use a unique ID so re-running the cell doesn't cause
# duplicate-ID errors.

persistent_collection.upsert(
    documents=[
        "This data will be saved to disk.",
        "ChromaDB can store vector embeddings locally."
    ],
    ids=[
        "p1",
        "p2"
    ]
)

print("✅ Data saved successfully!")
print("Storage location: /content/chroma_db")
print("Persistent collection count:", persistent_collection.count())


# ------------------------------------------------------------
# 19. SEARCH PERSISTENT DATABASE
# ------------------------------------------------------------
query = "Where is the vector database stored?"

results = persistent_collection.query(
    query_texts=[query],
    n_results=2
)

print("\nQuery:", query)
print("\nResults:")

for doc, distance in zip(results["documents"][0], results["distances"][0]):
    print(f"Distance: {distance:.4f} | {doc}")


# ------------------------------------------------------------
# 20. FINAL SUMMARY
# ------------------------------------------------------------
print("\n" + "=" * 60)
print("  CHROMADB DEMO COMPLETED SUCCESSFULLY!")
print("=" * 60)

print("""
You learned:

1. ✅ Installed ChromaDB
2. ✅ Created an in-memory ChromaDB client
3. ✅ Created a collection
4. ✅ Added documents
5. ✅ Performed semantic search
6. ✅ Added metadata
7. ✅ Filtered using metadata
8. ✅ Used Hugging Face embeddings
9. ✅ Updated documents
10. ✅ Deleted documents
11. ✅ Created persistent storage
12. ✅ Queried the persistent database

You now have the basic building blocks needed for
RAG (Retrieval-Augmented Generation) applications.
""")
```

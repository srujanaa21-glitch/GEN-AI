```python
# ============================================================
# ChromaDB CRUD Operations - Complete Google Colab Example
# ============================================================

# 1. INSTALL CHROMADB
!pip install -q chromadb


# 2. IMPORT AND CREATE CLIENT
import chromadb

# In-memory client
# Data will be lost when the Colab runtime restarts.
client = chromadb.Client()

# Create collection
collection = client.create_collection(name="crud_demo")

print("=" * 60)
print("1. CREATE - COLLECTION")
print("=" * 60)
print("Collection created:", collection.name)
print("Current document count:", collection.count())


# ============================================================
# 3. CREATE - ADD DOCUMENTS
# ============================================================

print("\n" + "=" * 60)
print("2. CREATE - ADD DOCUMENTS")
print("=" * 60)

collection.add(
    documents=[
        "Python is a popular programming language for AI.",
        "The weather today is sunny and warm.",
        "Machine learning models learn patterns from data.",
        "I enjoy playing football on weekends.",
        "Deep learning uses neural networks with many layers."
    ],
    metadatas=[
        {"category": "tech"},
        {"category": "weather"},
        {"category": "tech"},
        {"category": "sports"},
        {"category": "tech"}
    ],
    ids=["doc1", "doc2", "doc3", "doc4", "doc5"]
)

print("Documents added!")
print("Total document count:", collection.count())


# ============================================================
# 4. READ - PEEK
# ============================================================

print("\n" + "=" * 60)
print("3. READ - PEEK")
print("=" * 60)

preview = collection.peek(limit=3)

for doc_id, doc in zip(preview["ids"], preview["documents"]):
    print(doc_id, "->", doc)


# ============================================================
# 5. READ - GET BY ID
# ============================================================

print("\n" + "=" * 60)
print("4. READ - GET BY ID")
print("=" * 60)

result = collection.get(ids=["doc1", "doc3"])

for doc_id, doc, meta in zip(result["ids"], result["documents"], result["metadatas"]):
    print(f"{doc_id} | {meta} | {doc}")


# ============================================================
# 6. READ - SEMANTIC SEARCH
# ============================================================

print("\n" + "=" * 60)
print("5. READ - SEMANTIC SEARCH")
print("=" * 60)

results = collection.query(
    query_texts=["Tell me about AI and neural networks"],
    n_results=2
)

for doc_id, doc, distance in zip(results["ids"][0], results["documents"][0], results["distances"][0]):
    print(f"{doc_id} | distance={distance:.4f} | {doc}")


# ============================================================
# 7. READ - SEMANTIC SEARCH + METADATA FILTER
# ============================================================

print("\n" + "=" * 60)
print("6. READ - SEARCH WITH METADATA FILTER")
print("=" * 60)

results = collection.query(
    query_texts=["something fun to do outside"],
    n_results=3,
    where={"category": "sports"}
)

for doc_id, doc, distance in zip(results["ids"][0], results["documents"][0], results["distances"][0]):
    print(f"{doc_id} | distance={distance:.4f} | {doc}")


# ============================================================
# 8. UPDATE - MODIFY EXISTING DOCUMENT
# ============================================================

print("\n" + "=" * 60)
print("7. UPDATE - MODIFY DOCUMENT")
print("=" * 60)

collection.update(
    ids=["doc4"],
    documents=["I enjoy playing football and basketball on weekends."],
    metadatas=[{"category": "sports", "updated": True}]
)

updated = collection.get(ids=["doc4"])

print("Updated document:")
print(updated["documents"])

print("Updated metadata:")
print(updated["metadatas"])


# ============================================================
# 9. UPSERT - ADD OR UPDATE
# ============================================================

print("\n" + "=" * 60)
print("8. UPSERT - ADD OR UPDATE")
print("=" * 60)

collection.upsert(
    documents=[
        # doc5 already exists -> UPDATE
        "Deep learning uses neural networks with many layers and GPUs.",
        # doc6 does not exist -> CREATE
        "The Eiffel Tower is located in Paris."
    ],
    ids=["doc5", "doc6"]
)

print("Total count after upsert:", collection.count())

upsert_result = collection.get(ids=["doc5", "doc6"])

for doc_id, doc in zip(upsert_result["ids"], upsert_result["documents"]):
    print(doc_id, "->", doc)


# ============================================================
# 10. DELETE - BY ID
# ============================================================

print("\n" + "=" * 60)
print("9. DELETE - BY ID")
print("=" * 60)

collection.delete(ids=["doc2"])

print("Deleted: doc2")
print("Total count after delete:", collection.count())

print("Remaining IDs:")
print(collection.get()["ids"])


# ============================================================
# 11. DELETE - BY METADATA FILTER
# ============================================================

print("\n" + "=" * 60)
print("10. DELETE - BY METADATA FILTER")
print("=" * 60)

# Delete every document where category = tech
collection.delete(where={"category": "tech"})

print("Deleted all documents with category='tech'")

print("Total count after category delete:")
print(collection.count())

print("Remaining IDs:")
print(collection.get()["ids"])


# ============================================================
# 12. FINAL STATE
# ============================================================

print("\n" + "=" * 60)
print("11. FINAL COLLECTION STATE")
print("=" * 60)

final_state = collection.get()

print("Final document count:", collection.count())

if final_state["ids"]:
    for doc_id, doc, meta in zip(final_state["ids"], final_state["documents"], final_state["metadatas"]):
        print(f"{doc_id} | {meta} | {doc}")
else:
    print("No documents remaining.")


# ============================================================
# 13. SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("12. CRUD SUMMARY")
print("=" * 60)

print("""
CREATE
  client.create_collection()
  collection.add()

READ
  collection.get()
  collection.peek()
  collection.query()

UPDATE
  collection.update()

UPSERT
  collection.upsert()

DELETE
  collection.delete()

Other useful method:
  collection.count()
""")

print("=" * 60)
print("CRUD DEMO COMPLETED SUCCESSFULLY!")
print("=" * 60)
```

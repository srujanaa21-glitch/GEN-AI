```python
import os
# ============================================
# WORD EMBEDDING USING ONE-HOT ENCODING
# ============================================

from sklearn.preprocessing import OneHotEncoder

# List of words
words = [['cat'], ['dog'], ['apple'], ['car']]

# Create OneHotEncoder
encoder = OneHotEncoder(sparse_output=False)

# Fit and transform the words
result = encoder.fit_transform(words)

# Display vocabulary
print("Words:", encoder.categories_[0])

# Display one-hot encoded vectors
print("\nEmbeddings:")
print(result)


# ============================================
# FIND SIMILAR WORDS USING WORD2VEC
# ============================================

# Install gensim in Google Colab
!pip install gensim -q

# Import Word2Vec
from gensim.models import Word2Vec

# Training sentences
sentences = [
    ["python", "is", "a", "language"],
    ["java", "is", "a", "language"],
    ["python", "is", "easy"],
    ["java", "is", "easy"],
    ["python", "is", "powerful"],
    ["java", "is", "powerful"]
]

# Create and train Word2Vec model
model = Word2Vec(
    sentences,
    vector_size=10,
    window=2,
    min_count=1,
    workers=1
)

# Find words most similar to "java"
print("\nWords similar to 'java':")
print(model.wv.most_similar("java", topn=3))

# Find words most similar to "python"
print("\nWords similar to 'python':")
print(model.wv.most_similar("python", topn=4))
```
